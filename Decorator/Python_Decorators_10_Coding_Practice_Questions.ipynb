## Q1 — Basic: Create Your First Decorator

**Problem statement**

Create a decorator named `announce` that prints `Function is starting` before the decorated function runs. Decorate a function named `greet` that prints `Hello, Python!`. Call `greet()`.

**Concepts tested:** Defining functions, passing a function to a decorator, returning a wrapper, applying `@announce`.

**Expected output**
```text
Function is starting
Hello, Python!
```

**Edge cases:** Check that the message appears before the original function's message.

````python
# Write your answer here
def announce(funct):
    def wrapping():
        print("Function is starting")
        funct()
    return wrapping
  
    

@announce
def greet():
    print("Hello, Python!")

greet()

````

## Q2 — Basic: Run Code Before and After a Function

**Problem statement**

Create a decorator named `around_call` that prints `Before function` before calling the decorated function and `After function` after it finishes. Apply it to a function named `show_message` that prints `Practice makes progress`.

**Concepts tested:** Wrapper functions, call order, decorator syntax.

**Expected output**
```text
Before function
Practice makes progress
After function
```

**Edge cases:** Make sure the after-message is not printed before the decorated function runs.

````python
# Write your answer here
def around_call(funct):
    def wrapp():
        print("Before function")
        funct()
        print("After function")
    return wrapp
@around_call
def show_message():
    print("Practice makes progress")
show_message()
````

## Q3 — Basic: Preserve the Function's Return Value

**Problem statement**

Create a decorator named `keep_result` that calls the decorated function and returns its result. Decorate a function named `multiply` that takes two numbers and returns their product. Call `multiply(4, 5)` and print the returned value.

**Concepts tested:** Function parameters, return values, wrapper return behavior.

**Sample input**
```text
4 5
```

**Expected output**
```text
20
```

**Edge cases:** Test with zero and with a negative number.

````python
# Write your answer here
def keep_result(funct):
    def wrapping(*args):
        return funct(*args)
    return wrapping

@keep_result
def multiply(a,b):
    return a*b

print(multiply(4,5))
````

## Q4 — Intermediate: Decorator That Accepts Function Arguments

**Problem statement**

Create a decorator named `show_arguments` that works with a decorated function receiving any number of positional arguments. It should print the arguments as a tuple, then call the function. Decorate `add_numbers`, which returns the sum of the numbers it receives.

**Concepts tested:** `*args`, wrapper functions, loops or built-in summation, return values.

**Sample call**
```python
add_numbers(2, 3, 4)
```

**Expected output**
```text
Arguments: (2, 3, 4)
Result: 9
```

**Edge cases:** Try calling the function with one argument and with no arguments.

````python
# Write your answer here
def show_arguments(func):
    def wrapping(*args):
        return func(*args)
    return wrapping

@show_arguments
def add_numbers(*args):
    s=0
    for i in args:
        s=s+i
    return s

add_numbers(2,3,4)

````

## Q5 — Intermediate: Check a Result Before Returning It

**Problem statement**

Create a decorator named `positive_result` for a function that returns a number. If the result is positive, return it unchanged. If the result is zero or negative, return `0`. Decorate a function named `calculate` that returns the number passed to it.

**Concepts tested:** Conditions, decorators, return values.

**Sample calls and expected outputs**
```text
calculate(8)   -> 8
calculate(0)   -> 0
calculate(-3)  -> 0
```

**Edge cases:** Test positive, zero, and negative results.

````python
# Write your answer here
def positive_result(funct):
    def wrapping(*args):
        r=funct(*args)
        if r>0:
            return r
        else:
            return 0
    return wrapping

@positive_result
def calculate(n):
    return n
print(calculate(8))   # 8
print(calculate(0))   # 0
print(calculate(-3))  # 0
````

## Q6 — Intermediate: Count How Many Times a Function Runs

**Problem statement**

Create a decorator named `count_calls` that counts how many times the decorated function is called. After each call, print the total call count. Decorate a function named `say_hi` that prints `Hi`. Call it three times.

**Concepts tested:** Decorators, variables captured by a wrapper, repeated function calls.

**Expected output**
```text
Hi
Call count: 1
Hi
Call count: 2
Hi
Call count: 3
```

**Edge cases:** The count should increase once per call, not when the function is decorated.

````python
# Write your answer here

def count_calls(funct):
    s=0
    def wrapping():
        nonlocal s
        s=s+1
        return funct(s)
    return wrapping

@count_calls
def say_hi(s):
    print("Hi")
    print(s)


say_hi()
say_hi()
say_hi()

````
