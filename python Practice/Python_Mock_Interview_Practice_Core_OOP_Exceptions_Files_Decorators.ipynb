## Q1 — Basic | Core Python: Count Vowels

**Problem statement**

Write a function `count_vowels(text)` that returns the number of vowels in a string. Treat uppercase and lowercase vowels equally.

**Concepts tested:** Strings, loops, conditions, functions

**Sample input/output**

```text
Input: `"Data Scientist"
Output: `5`
```

**Constraints / edge cases:** An empty string should return `0`.

````python
# Write your answer here
s="Data Scientist"

def count_vowels(s):
    c=0
    for i in s:
        if i.lower() in ['a','e','i','o','u']:
            c=c+1
    return c



count_vowels(s)

````

## Q2 — Basic | Core Python: Separate Even and Odd Numbers

**Problem statement**

Write a function `separate_even_odd(numbers)` that returns two lists: the even numbers and the odd numbers, preserving their original order.

**Concepts tested:** Lists, loops, modulo, return values

**Sample input/output**

```text
Input: `[4, 7, 2, 9, 10]`
Output: `([4, 2, 10], [7, 9])`
```

**Constraints / edge cases:** An empty list should return two empty lists.

````python
# Write your answer here
def separate_even_odd(n):
    el=[]
    ol=[]
    for i in n:
        if i%2==0:
            el.append(i)
        else:
            ol.append(i)
    return el,ol

separate_even_odd([4,7,2,9,10])

````

## Q3 — Basic | Core Python: Build a Frequency Dictionary

**Problem statement**

Write a function `character_frequency(text)` that returns a dictionary containing each character and its count. Ignore spaces and treat uppercase/lowercase as the same character.

**Concepts tested:** Strings, dictionaries, loops

**Sample input/output**

```text
Input: `"Data"`\nOutput: `{'d': 1, 'a': 2, 't': 1}`
```

**Constraints / edge cases:** Do not count spaces.

````python
# Write your answer here
t="Data "

def character_frequency(t):
    t=t.lower()
    d={}
    
    for i in t:
        
        if i in d:
            d[i]=d[i]+1
        elif i==" ":
            continue
        else:
            d[i]=1
    return d

    


character_frequency(t)
            


````

## Q5 — Basic | Lists: Remove Duplicates

**Problem statement**

Write a function `unique_values(items)` that returns a list with duplicates removed while preserving the first occurrence of each item.

**Concepts tested:** Lists, membership checks, loops

**Sample input/output**

```text
Input: `[3, 1, 3, 2, 1]`\nOutput: `[3, 1, 2]`
```

**Constraints / edge cases:** Do not sort the result.

````python
# Write your answer here
def unique_values(a):
    l=[]
    for i in a:
        if i not in l:
            l.append(i)
    return l

unique_values([3,1,3,2,1])

````

## Q9 — Advanced | Core Python: Group Words by Length

**Problem statement**

Write a function `group_by_length(words)` that returns a dictionary where each key is a word length and each value is a list of words of that length. Preserve the input order within each group.

**Concepts tested:** Lists, dictionaries, loops

**Sample input/output**

```text
Input: `["cat", "to", "dog", "a", "sun"]`
Output: `{3: ['cat', 'dog', 'sun'], 2: ['to'], 1: ['a']}`
```

**Constraints / edge cases:** An empty input should return an empty dictionary.

````python
# Write your answer here
def group_by_length(w):
    d={}
    
    for i in w:
        
        if len(i) in d:
            d[len(i)].append(i)
        else:
            d[len(i)]=[i]
    return d




w=["cat", "to", "dog", "a", "sun"]
group_by_length(w)
````

## Q13 — Intermediate | OOP / Magic Method: Compare Objects

**Problem statement**

Create a class `Candidate` with `name` and `score`. Implement `__lt__` so two candidates can be compared using `<` based on their scores.

**Concepts tested:** Objects, magic method `__lt__`, comparisons

**Sample input/output**

```text
Input: candidate A score `70`, candidate B score `85`\nOutput of `candidate_a < candidate_b`: `True`
```

**Constraints / edge cases:** Compare scores, not names.

````python
# Write your answer here
class Candidate:
    def __init__(self,name,score):
        self.name=name
        self.score=score
    def __lt__(self,other):
        if self.score<other.score:
            return True
        else:
            return False
c1=Candidate("x",70)
c2=Candidate("y",15)
print(c1>c2)
````
