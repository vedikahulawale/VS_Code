## Q1. Basic — Abstract Shape

**Level:** Basic  
**Main concept:** ABC, @abstractmethod, implementing a method in a child class

Create an abstract class `Shape` with an abstract method `area()`.

Create a class `Rectangle` that inherits from `Shape`. Its constructor should accept `length` and `width`, and `area()` should return `length * width`.

**Requirements**
- `Shape` must inherit from `ABC`.
- Mark `area()` with `@abstractmethod`.
- Create a `Rectangle(5, 3)` object and print its area.

**Expected output**
```text
15
```

**Critical-thinking check:** What happens if you try to create `Shape()` directly, and why?

````python
from abc import ABC, abstractmethod

# Write your solution below
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
class Rectangle(Shape):

    def __init__(self,length,width):
        self.length=length
        self.width=width

    def area(self):
        a=self.length*self.width
        return a
r=Rectangle(5,3)
r.area()
````

## Q2. Basic — Multiple Implementations

**Level:** Basic  
**Main concept:** One abstract interface, different child-class implementations

Create an abstract class `Animal` with an abstract method `sound()`.

Create two child classes:
- `Dog`: `sound()` returns `"Bark"`
- `Cat`: `sound()` returns `"Meow"`

Create one object of each class and print the result of calling `sound()`.

**Expected output**
```text
Bark
Meow
```

**Critical-thinking check:** Both classes have the same method name. Why can they return different results?

````python
from abc import ABC, abstractmethod

# Write your solution below
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Bark"
class Cat(Animal):
    def sound(self):
        return "Meow"

d=Dog()
print(d.sound())
c=Cat()
print(c.sound())

````

## Q3. Basic–Intermediate — Payment System

**Level:** Basic–Intermediate  
**Main concept:** Using abstraction to define a required operation

Create an abstract class `Payment` with an abstract method `pay(amount)`.

Create these child classes:
- `CardPayment`: prints `Paid ₹<amount> using card`
- `UPIPayment`: prints `Paid ₹<amount> using UPI`

Create each payment object and call `pay(500)`.

**Expected output**
```text
Paid ₹500 using card
Paid ₹500 using UPI
```

**Critical-thinking check:** Why is it useful for every payment type to provide its own `pay()` implementation?

````python
from abc import ABC, abstractmethod

# Write your solution below
class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        self.amount=amount
class CardPayment(Payment):
    def pay(self,amount):
        self.amount=amount
        print(f"Paid ₹{self.amount} using card")
class UPIPayment(Payment):
    def pay(self,amount):
        self.amount=amount
        print(f"Paid ₹{self.amount} using UPI")
c=CardPayment()
c.pay(500)
u=UPIPayment()
u.pay(500)



````

## Q4. Intermediate — Employee Salary Calculator

**Level:** Intermediate  
**Main concept:** Abstract methods with child-specific business logic

Create an abstract class `Employee` with:
- An initializer that stores `name`
- An abstract method `calculate_salary()`

Create two child classes:
- `FullTimeEmployee(name, monthly_salary)`: returns the monthly salary
- `PartTimeEmployee(name, hours_worked, hourly_rate)`: returns `hours_worked * hourly_rate`

Create one object of each class and print each employee's name and calculated salary.

**Example**
A part-time employee who works 20 hours at ₹300 per hour earns ₹6000.

**Critical-thinking check:** Why should `calculate_salary()` be required in the base class instead of leaving it as an ordinary empty method?

````python
from abc import ABC, abstractmethod

# Write your solution below

class Employee(ABC):
    def __init__(self,name):
        self.name=name
    @abstractmethod
    def calculate_salary(self):
        pass


class FullTimeEmployee(Employee):
    def __init__(self,name,monthly_salary):
        super().__init__(name)
        self.monthly_salary=monthly_salary
    def calculate_salary(self):
        return self.monthly_salary

    
class PartTimeEmployee(Employee):
    def __init__(self,name,hours_worked,hourly_rate):
        super().__init__(name)
        self.hours_worked=hours_worked
        self.hourly_rate=hourly_rate
    def calculate_salary(self):
        return self.hours_worked*self.hourly_rate

f=FullTimeEmployee("vedika",100000)
print(f"Employee Name:{f.name} and Salary:{f.calculate_salary()}")

p=PartTimeEmployee("xy",20,300)
print(f"Employee Name:{p.name} and Salary:{p.calculate_salary()}")

````

## Q5. Intermediate — Notification Service

**Level:** Intermediate  
**Main concept:** Abstract interfaces and consistent method contracts

Create an abstract class `Notification` with an abstract method `send(message)`.

Implement:
- `EmailNotification`: prints `Email: <message>`
- `SMSNotification`: prints `SMS: <message>`
- `PushNotification`: prints `Push: <message>`

Store one object of each class in a list. Loop through the list and call `send("Your order is ready")` on each object.

**Expected output**
```text
Email: Your order is ready
SMS: Your order is ready
Push: Your order is ready
```

**Critical-thinking check:** How does this design make it easier to add a new notification type later?

````python
from abc import ABC, abstractmethod

# Write your solution below
class Notification(ABC):
    
    @abstractmethod
    def send(self,msg):
        self.msg=msg
    
class Email(Notification):

    def send(self,msg):
        self.msg=msg
        return f"Email:{self.msg}"
    
class SMSNotification(Notification):

    def send(self,msg):
        self.msg=msg
        return f"SMS:{self.msg}"

class PushNotification(Notification):

    def send(self,msg):
        self.msg=msg
        return f"Push:{self.msg}"
e=Email()
print(e.send("Your order is ready"))

s=SMSNotification()
print(s.send("Your order is ready"))

p=PushNotification()
print(p.send("Your order is ready"))
        
````

## Q6. Intermediate — Data Exporter

**Level:** Intermediate  
**Main concept:** Abstract method contracts and format-specific implementation

Create an abstract class `DataExporter` with an abstract method `export(data)`.

Implement two child classes:
- `CSVExporter`: returns a string containing comma-separated values from a list of strings.
- `TextExporter`: returns the list items joined with a newline character.

For `data = ["name", "age", "city"]`, print the result from each exporter.

**Expected CSV output**
```text
name,age,city
```

**Expected text output**
```text
name
age
city
```

**Critical-thinking check:** The output formats differ. What remains consistent about the interface?

````python
from abc import ABC, abstractmethod

# Write your solution below
class DataExporter(ABC):

    @abstractmethod
    def export(self,data):
        self.data=data

class CSVExporter(DataExporter):

    def export(self, data):
        self.data=data
        for i in self.data:
            print(i,end="")
            print(",",end="")

class TextExporter(DataExporter):

    def export(self, data):
        self.data=data
        for i in self.data:
            print(i)
c=CSVExporter()
print(c.export(["name","age","city"]))

t=TextExporter()
t.export(["name","age","city"])

    

````

## Q7. Intermediate–Advanced — Appliance Controller

**Level:** Intermediate–Advanced  
**Main concept:** Abstract actions and state changes

Create an abstract class `Appliance` with abstract methods `turn_on()` and `turn_off()`.

Implement a `Fan` class that stores an `is_on` state:
- Initially, the fan is off.
- `turn_on()` changes the state to on and prints `Fan is ON`.
- `turn_off()` changes the state to off and prints `Fan is OFF`.

Create a fan, turn it on, then turn it off.

**Expected output**
```text
Fan is ON
Fan is OFF
```

**Critical-thinking check:** What error would occur if `Fan` implemented `turn_on()` but forgot `turn_off()`?

````python
from abc import ABC, abstractmethod

# Write your solution below
class Application(ABC):
    @abstractmethod
    def turn_on(self):
        pass
    @abstractmethod
    def turn_off(self):
        pass

class Fan(Application):

    def __init__(self,is_on):
        self.is_on=is_on

    def turn_on(self,is_on):
        self.is_on=is_on
        return f"Fan is {self.is_on}"
    
    def turn_off(self,is_on):
        self.is_on=is_on
        return f"Fan is {self.is_on}"

f=Fan("off")
print(f.turn_on("ON"))
print(f.turn_off("OFF"))

````
