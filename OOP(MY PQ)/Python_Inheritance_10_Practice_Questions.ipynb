## Q1. Basic Single Inheritance

**Concept:** A child class can reuse attributes and methods from a parent class.

Create a parent class `Animal` with:
- `__init__(self, name)` that stores `name`
- `eat()` that prints `"<name> is eating"`

Create a child class `Dog` that inherits from `Animal` and adds a method `bark()` that prints `"<name> is barking"`.

Create `Dog("Bruno")`, then call both methods.

**Expected output**
```text
Bruno is eating
Bruno is barking
```

````python
# Write your solution here
class Animal:
    def __init__(self,name):
        self.name=name
    def eat(self):
        return f"{self.name} is eating"
class Dog(Animal):
    def bark(self):
        return f"{self.name} is barking"

d=Dog("Bruno")
print(d.eat())
print(d.bark())
````

## Q2. Child Constructor and `super()`

**Concept:** Use `super()` to call the parent class constructor and avoid repeating parent setup.

Create a parent class `Person` with `name` and `age`. Create a child class `Student` with `name`, `age`, and `grade`.

Requirements:
- `Student.__init__()` must use `super()` to initialize `name` and `age`.
- Add `display()` to print all three values.

Test with `Student("Asha", 20, "A")`.

**Expected output**
```text
Name: Asha
Age: 20
Grade: A
```

````python
# Write your solution here
class Person:
    def __init__(self,name,age):
        self.age=age
        self.name=name
class Student(Person):
    def __init__(self,name,age,grade):
        super().__init__(name,age)
        self.grade=grade
        
        
    def display(self):
        print(f"Name:{self.name}\nAge:{self.age}\nGrade:{self.grade}")
                
s=Student("vedika",23,100)
s.display()

````

## Q3. Inheriting a Constructor Without Defining One

**Concept:** If a child class does not define `__init__()`, Python can use the inherited parent constructor.

Create a class `Vehicle` whose constructor accepts `brand` and stores it. Add `show_brand()` to print the brand.

Create an empty child class `Car(Vehicle)` with no constructor of its own. Create `Car("Tata")` and call `show_brand()`.

**Expected output**
```text
Brand: Tata
```

````python
# Write your solution here
class Vehicle:
    def __init__(self,brand):
        self.brand=brand

    def show_brand(self):
        print(self.brand)
class Car(Vehicle):
    pass

c=Car("Tata")
c.show_brand()

````

## Q4. Method Overriding

**Concept:** A child class can provide its own version of an inherited method.

Create a parent class `Employee` with a method `calculate_bonus()` that returns `1000`.

Create a child class `Manager` that overrides `calculate_bonus()` and returns `5000`.

Print the bonus for an `Employee` object and a `Manager` object.

**Expected output**
```text
Employee bonus: 1000
Manager bonus: 5000
```

````python
# Write your solution here
class Employee:
    def __init__(self,bonus):
        self.bonus=bonus
    def calculate_bonus(self,bonus):
        self.bonus=bonus
        return f"Employee bonus:{bonus}"

class Manager(Employee):
    def __init__(self,bonus):
        self.bonus=bonus
    def calculate_bonus(self, bonus):
        self.bonus=bonus
        print( f"{super().calculate_bonus(1000)}\nManager bonus:{self.bonus}")
    
m=Manager(1000)
m.calculate_bonus(5000)
````

## Q5. Call the Parent Method Using `super()`

**Concept:** `super()` can call a parent method even when the child overrides that method.

Create a parent class `Notification` with `send()` printing `Sending a general notification`.

Create child class `EmailNotification` that overrides `send()`:
1. Call the parent `send()` using `super()`.
2. Print `Sending an email`.

Call `send()` on an `EmailNotification` object.

**Expected output**
```text
Sending a general notification
Sending an email
```

````python
# Write your solution here
class Notification:
    def send(self):
        print("Sending a general notification")
class EmailNotification(Notification):
    def send(self):
        super().send()
        print("Sending an email")
e=EmailNotification()
e.send()
````

## Q6. Multilevel Inheritance

**Concept:** In multilevel inheritance, a class inherits from a child class, creating a chain.

Create:
- `Person` with a method `introduce()` that prints `I am a person`
- `Employee(Person)` with `work()` that prints `I am working`
- `DataScientist(Employee)` with `analyze_data()` that prints `I analyze data`

Create a `DataScientist` object and call all three methods.

**Expected output**
```text
I am a person
I am working
I analyze data
```

````python
# Write your solution here
class Person:
    def introduce(self):
        print("I am a person")
class Employee(Person):
    def work(self):
        print("I am working")
class DataScientist(Employee):
    def analyze_data(self):
        print("I analyze data")

d=DataScientist()
d.introduce()
d.work()
d.analyze_data()


````

## Q7. Hierarchical Inheritance

**Concept:** Multiple child classes can inherit from the same parent class.

Create a parent class `Shape` with `describe()` printing `I am a shape`.

Create two child classes:
- `Circle` with `draw()` printing `Drawing a circle`
- `Rectangle` with `draw()` printing `Drawing a rectangle`

Create one object of each child class. Call `describe()` and `draw()` for both.

**Expected output**
```text
I am a shape
Drawing a circle
I am a shape
Drawing a rectangle
```

````python
# Write your solution here
class Shape:
    def describe(self):
        return "I am a shape"
class Circle(Shape):
    def draw(self):
        return "Drawing a circle"
class Rectangle(Shape):
    def draw(self):
        return "Drawing a rectangle"

c=Circle()
print(c.describe())
print(c.draw())
r=Rectangle()
print(r.describe())
print(r.draw())

````

## Q8. Multiple Inheritance

**Concept:** A child class can inherit from more than one parent class.

Create:
- `Camera` with `take_photo()` printing `Photo captured`
- `Phone` with `make_call()` printing `Calling...`
- `SmartPhone(Camera, Phone)` with no extra methods

Create a `SmartPhone` object and call both inherited methods.

**Expected output**
```text
Photo captured
Calling...
```

````python
# Write your solution here
class Camera:
    def take_photo(self):
        return "Photo Captured"

class Phone:
    def make_call(self):
        return "Calling.."

class SmartPhone(Camera,Phone):
    pass

s=SmartPhone()
print(s.take_photo())
print(s.make_call())

````

## Q9. Method Resolution Order (MRO)

**Concept:** When multiple parents define the same method, Python follows the Method Resolution Order (MRO).

Create:
- `A` with `show()` printing `A`
- `B(A)` overriding `show()` to print `B`
- `C(A)` overriding `show()` to print `C`
- `D(B, C)` with no `show()` method of its own

Create `D()` and call `show()`. Then print the class names in `D.mro()`.

**Expected output**
```text
B
['D', 'B', 'C', 'A', 'object']
```

**Think about it:** Why does Python choose `B.show()` first?

````python
# Write your solution here
class A:
    def show(self):
        print("A")
class B(A):
    def show(self):
        print("B")
class C(A):
    def show(self):
        print("C")
class D(B,C):
    pass

d=D()
d.show()
D.mro()
````

## Q10. Mini Project — Employee Payroll

**Concept:** Combine inheritance, constructors, `super()`, and method overriding.

Create a parent class `Employee`:
- Constructor accepts `name` and `base_salary`.
- `calculate_salary()` returns `base_salary`.

Create two child classes:
- `Developer(Employee)`: constructor also accepts `bonus`; use `super()` for parent attributes. Override `calculate_salary()` to return base salary + bonus.
- `Intern(Employee)`: constructor also accepts `stipend`; use `super()` for parent attributes. Override `calculate_salary()` to return the stipend.

Create:
- `Developer("Ravi", 50000, 8000)`
- `Intern("Mina", 10000, 12000)`

Print each employee's name and calculated salary.

**Expected output**
```text
Ravi salary: 58000
Mina salary: 12000
```

**Extra challenge:** Add a third child class `TeamLead` with a `leadership_allowance`, and calculate its salary using the base salary plus the allowance.

````python
# Write your solution here
class Employee:
    def __init__(self,name,base_salary):
        self.name=name
        self.base_salary=base_salary
    def calculate_salary(self):
        return self.base_salary
    
class Developer(Employee):

    def __init__(self,name,base_salary,bonus):
        self.bonus=bonus
        super().__init__(name,base_salary)

    def calculate_salary(self):
        return self.base_salary+self.bonus

class Intern(Employee):

    def __init__(self,name,base_salary,stipend):
        super().__init__(name,base_salary)
        self.stipend=stipend

    def calculate_salary(self):
        return self.stipend

class TeamLead(Employee):
    def __init__(self,name,base_salary,leadership_allowance):
        self.leadership_allowance=leadership_allowance
        super().__init__(name,base_salary)

    def calculate_salary(self):
        return self.base_salary+self.leadership_allowance



d=Developer("Ravi", 50000, 8000)
print(d.calculate_salary())

i=Intern("Mina", 10000, 12000)
print(i.calculate_salary())

t=TeamLead("yeva", 1000000, 120000)
print(t.calculate_salary())
````
