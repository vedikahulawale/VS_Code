## Q1 — Basic: Create and write to a file
Write a program that creates a text file named `welcome.txt` and writes these two lines:
- `Welcome to Python`
- `File handling practice`

**Concepts tested:** `open()`, write mode (`w`), `write()`, newline characters, closing a file.

**Expected result:** `welcome.txt` contains exactly:
```text
Welcome to Python
File handling practice
```

**Edge case:** Run the program twice. Observe what happens to the existing file contents.

````python
f=open("welcome.txt","w")
f.write("welcome to python\n")
f.write("File handling practice")
f.close()
````

````python
# Write your solution here
with open("welcome.txt","w") as f:
    f.write("welcome to python\n")
    f.write("File handling practice")
````

## Q4 — Intermediate: Count lines in a file
Create a file named `notes.txt` with the following content:
```text
Python
SQL
Statistics
Machine Learning
```
Read the file and print the number of lines.

**Concepts tested:** reading lines, loops, counting.

**Sample output**
```text
Number of lines: 4
```

**Edge cases:** Test a file with one line and an empty file.

````python
f=open("notes.txt","r")
while True:
    d=f.read()
    if d!="":
        print(d)
    else:
        break
````

````python
# Write your solution here
l=["Python\n","SQL\n","Statistics\n","Machine Learning\n"]
with open("notes.txt","w") as f:
    f.writelines(l)
with open("notes.txt","r") as f:
    print(f.read())
````

## Q5 — Intermediate: Count words in a file
Create `article.txt` with this content:
```text
Python is useful
Python is readable
Data needs practice
```
Read the file and count how many words it contains. Treat words separated by whitespace as individual words.

**Concepts tested:** file reading, loops or string methods, counters.

**Sample output**
```text
Word count: 10
```

**Edge cases:** Include multiple spaces and blank lines. Do not count blank spaces as words.

````python
# Write your solution here

with open("article.txt", "w") as f:
    f.write("Python is useful\nPython is readable\nData needs practice")

with open("article.txt", "r") as f:
    d = f.read()
    words = d.split()

print("Word count:", len(words))

````
