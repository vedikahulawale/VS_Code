## Q1 — Basic: Handle invalid number input

**Problem statement:**  
Write a program that asks the user to enter an integer. If the input cannot be converted to an integer, display `Invalid input`. Otherwise, display the number multiplied by 2.

**Concepts tested:** `try`, `except`, `ValueError`, integer conversion.

**Sample input**
```text
Enter an integer: 12
```

**Sample output**
```text
24
```

**Another test**
```text
Enter an integer: hello
```
Expected output:
```text
Invalid input
```

**Edge cases:** Negative integers, zero, and text that is not a valid integer.

````python
# Write your answer for Q1 here
i=input("Enter the value")
try:
    i=int(i)
except ValueError:
    print("Invalid input")
else:
    r=i*2
    print(r)


````

## Q2 — Basic to Intermediate: Safe division

**Problem statement:**  
Ask the user to enter two numbers and divide the first number by the second. Handle invalid numeric input and division by zero with clear messages. If the division succeeds, display the result.

**Concepts tested:** Handling more than one exception, `ValueError`, `ZeroDivisionError`.

**Sample input**
```text
Enter numerator: 10
Enter denominator: 0
```

**Sample output**
```text
Cannot divide by zero
```

**Edge cases:** Non-numeric input, a negative denominator, and a valid division that produces a decimal result.

````python
# Write your answer for Q2 here
try:
    a=int(input("Enter the 1st number "))
    b=int(input("Enter the 2nd number "))
    r=a/b
except ValueError:
    print("Invalid input")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print(r)

````

## Q3 — Intermediate: Process a list of numeric strings

**Problem statement:**  
Given a list of strings, convert each valid integer string to an integer and add it to a total. If an item is invalid, skip that item and continue processing the remaining items. Display the final total and the number of invalid items.

**Concepts tested:** `try`/`except` inside a loop, `ValueError`, counters, accumulators, list processing.

**Sample input**
```python
values = ["10", "5", "hello", "-2", "3x"]
```

**Sample output**
```text
Total: 13
Invalid items: 2
```

**Edge cases:** An empty list, all items invalid, negative numbers, and strings containing spaces.

````python
# Write your answer for Q3 here
v=['10','5','hello','-2','3x']

c=0
s=0
for i in v:
    try:
        s=int(i)+s
    except ValueError:
        c=c+1
else:
    print("Total:",s)
print("Invalid items:",c)

    

````

## Q4 — Advanced: Validate a record and report errors

**Problem statement:**  
A data record is provided as a dictionary with the keys `name`, `age`, and `score`. Write a program that checks whether:
- `name` exists and is not an empty string after removing surrounding spaces.
- `age` can be converted to an integer and is at least 18.
- `score` can be converted to a number (`float`).

Handle missing keys and invalid conversions without crashing. Collect and display all validation error messages you find, rather than stopping after the first error. If there are no errors, display `Record is valid`.

**Concepts tested:** `try`/`except`, `KeyError`, `ValueError`, conditions, dictionaries, collecting multiple errors.

**Sample input**
```python
record = {"name": "Asha", "age": "21", "score": "87.5"}
```

**Sample output**
```text
Record is valid
```

**Another test**
```python
record = {"name": " ", "age": "seventeen"}
```
Expected output: show errors for the blank name, invalid age, and missing `score`.

**Edge cases:** Missing keys, blank names, age below 18, and non-numeric scores.

````python
# Write your answer for Q4 here
r = {"name": "Asha", "age": "21", "score": "87.5"}
l=[]
try:
    if r["name"].strip()=="":
        l.append("Name cannot be blank")
except KeyError:
    l.append("name not found")       


    
try:
    i=int(r["age"])
    if i<18:
        l.append("Age must be at least 18")

except ValueError:
    l.append("Invalid age")
except KeyError:
    l.append("age not found")

    

    
try:
    if r["score"].strip()=="":
        l.append("Score cannot be blank")
    else:
        float(r["score"])
except ValueError:
    l.append("Invalid score")
except KeyError:
    l.append("Score is missing")




    

    
if len(l)==0:
    print("Record is valid")
else:
    print(l)
````

## Q5 — Advanced / Critical Thinking: Use `else` and `finally`

**Problem statement:**  
Write a program that asks for a filename and tries to open the file and read its contents. If the file cannot be opened because it does not exist, display `File not found`. If it opens successfully, display the file contents and a message saying `Read successful`. Use `else` for the success-only message and `finally` to display `Operation finished` whether the read succeeds or fails. Make sure the program does not try to read from a file that failed to open.

**Concepts tested:** `try`, `except`, `else`, `finally`, `FileNotFoundError`, safe file handling.

**Sample scenario A — existing file containing `Hello Python`**

**Expected output**
```text
Hello Python
Read successful
Operation finished
```

**Sample scenario B — filename does not exist**

**Expected output**
```text
File not found
Operation finished
```

**Edge cases:** Empty files, files with multiple lines, and a filename that does not exist.

````python
# Write your answer for Q5 here
try:
    with open("sample.txt","r") as f:
        print(f.read())


except FileNotFoundError:
    print("File not found")
else:
    print("Read successful")

finally :
    print("Operation finished")
````
