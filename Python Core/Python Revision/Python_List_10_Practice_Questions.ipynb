### Q1. Sum of all elements

Given a list of numbers, calculate and print the sum of all elements.

Example:
```python

```

### Q6. Separate even and odd numbers

From a given list, create two lists: one containing even numbers and the other containing odd numbers. Print both lists.

Example:
```python
numbers = [1, 2, 3, 4, 5, 6]
```
Expected output: `Even: [2, 4, 6]`, `Odd: [1, 3, 5]`

````python
# Write your solution here
numbers = [1, 2, 3, 4, 5, 6]
e=[]
o=[]

for i in numbers:
    if i%2==0:
        e.append(i)
    else:
        o.append(i)

print(f"even:{e},odd:{o}")
````

### Q7. Find pairs with a target sum

Given a list of integers and a target value, find all unique pairs whose sum equals the target. Avoid printing the same pair more than once.

Example:
```python
numbers = [2, 4, 3, 5, 7, 8, 3]
target = 10
```
Expected pairs: `(2, 8)`, `(3, 7)`, `(5, 5)` if the input contains two 5s; otherwise include only pairs supported by the input.

````python
# Write your solution here
numbers = [2, 4, 3, 5, 7, 8, 3]
l=[]
for i in numbers:
    for j in numbers:
        if i+j==10:
            if (i,j) not in l and (j,i) not in l:
                l.append((i,j))
l      

````

### Q10. Find the most frequent element

Given a list, find the element that appears most frequently and its frequency. If multiple elements share the highest frequency, return all of them.

Example:
```python
values = [4, 2, 4, 3, 2, 4, 2]
```
Expected output: Most frequent elements `[4, 2]`, frequency `3`

````python
# Write your solution here
values = [4, 2, 4, 3, 2, 4, 2]
d={}
for i in values:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
mc=0
for i,j in d.items():
    if j>mc:
        mc=j
l=[]
for i,j in d.items():
    if j==mc:
        l.append(i)

l
````
