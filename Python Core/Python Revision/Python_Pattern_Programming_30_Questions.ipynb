### Q7. Alphabet Triangle


### Q11. Right-Aligned Triangle

Print the following pattern using Python:

```text
      *
    * *
  * * *
* * * *
```

**Your code:**

````python
# Write your solution here
for i in range(1,5):
    print(" "*((4-i)*2),end="")
    for k in range(1,i+1):
        print("* ",end="")
    print()
````
