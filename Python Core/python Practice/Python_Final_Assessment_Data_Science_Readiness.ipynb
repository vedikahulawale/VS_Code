### Q19. Create a custom iterator

Create a class `CountDown` that:
- Accepts a starting number.
- Returns numbers in descending order.
- Stops at zero.
- Implements `__iter__()` and `__next__()`.

**Expected output for starting number 5**
```python
5
4
3
2
1
```

````python
# Write your answer here
class CountDown:
    def __init__(self,n):
        self.n=n
    def __iter__(self):
        return self

    def __next__(self):
        if self.n>0:
            r=self.n
            self.n-=1
            return r
        else:
            raise StopIteration

        
c=CountDown(5)

for i in c:
    print(i)
````

### Q20. Implement a decorator to calculate execution count

Create a decorator that counts how many times a function is called.

**Expected output**
```text
Function called 1 times
Function called 2 times
Function called 3 times
```

````python
# Write your answer here
def decorator(func):
    c=0
    def wrapping():
        nonlocal c
        c+=1
        print(f"Function called {c} times")
        func()
    return wrapping

@decorator
def fun():
    pass
    
fun() 
fun() 

````
