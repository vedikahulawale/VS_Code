### Q1 — Basic: Even or Odd Using List Comprehension

Given a list of integers, use **list comprehension** to create a new list that classifies each number as `"Even"` or `"Odd"`.

**Sample input**
```python

```

**Expected output**
```python

```

**Concepts tested:** List comprehension, conditional expression, modulus operator.

````python
# Write your answer here
n=[10, 15, 20, 7, 8, 13]
l=[]
for i in n:
    if i%2==0:
        l.append("Even")
    else:
        l.append("Odd")
print(l)
        
````

### Q3 — Basic: Find the Maximum Number in a List

Given a list of integers, find the largest number. Use list comprehension to generate the required values.

**Conditions**
- Do not use `max()`.

```python

```

**Expected output**
```text

```

**Concepts tested:** List comprehension, conditions, comparisons, list traversal.

````python
# Write your answer here

#using max function 
n=[12, 45, 7, 89, 23, 56]
print(max(n))

#for loop 
m=n[0]
for i in n:
    if m<i:
        m=i
print(m)

#list comprehension
l = [i for i in n if all(i >= j for j in n)]
print(l)
````

**Conditions**
- Do not use `Counter` or `max()`.
- Use a dictionary to count frequencies.
- If multiple elements share the highest frequency, display all of them.
- Handle an empty list.

**Sample input**
```python

```

**Expected output**
```text
Most frequent element: 10
Frequency: 4
```

**Concepts tested:** Dictionaries, frequency counting, loops, conditions, tie handling.

````python
# Write your answer here
n=[10, 20, 10, 30, 20, 10, 40, 20, 10]

#using counter method
l=[]
for i in n:
    if i not in l:
        l.append(i)
d={}
# store in dictonary 
for i in l:
    d[i]=n.count(i)

#max count
m=0
for k,v in d.items():
    if m<v:
        m=v


#find key
for k,v in d.items():
    if v==m:
        mk=k
print(f"{mk}:{m}")


#using for loop 
d={}
n=[10, 20, 10, 30, 20, 10, 40, 20, 10]

for i in n:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
m=0
for k,v in d.items():
    if m<v:
        m=v

for k,v in d.items():
    if m==v:
        km=k

print(f"{km}:{m}")

````
